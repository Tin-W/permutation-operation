## Part 1: the Permutation class

Most of the code you write in this project will be in the next cell, which will contain the definition of your `Permutation` class.  This class will represent a permutation of the numbers $1, 2, 3, \ldots, n$ for some $n \geqslant 1$.  People who want to create a `Permutation` corresponding to $\sigma \in S_n$ will have to supply a list of the $n$ numbers on the bottom row of the two row notation for $\sigma$ - re-read the [Algebra 1 lecture notes](https://www.ucl.ac.uk/~ucahmto/0005_2023/Ch2.S10.html) if you've forgotten what this means.  For example, if we do

```python
a = Permutation([1, 2, 3, 4])
b = Permutation([2, 3, 1])
c = Permutation([2, 1, 4, 3])
```

then `a` will be the identity permutation in $S_4$, `b` will be $(1, 2, 3) \in S_3$, and $c$ will be $(1, 2)(3, 4) \in S_4$.

**Write all of the following methods for the Permutation class.**  I've listed these in roughly increasing order of difficulty.

1. `__str__(self)` must return a string that represents the bottom row of the permutation. Space the numbers out somehow to make them easy to read.
2. `__eq__(self, other)` should return `True` if self and other represent the same permutation and `False` otherwise.  This is easy - you just have to check whether they have the same bottom row.
3. `__mul__(self, other)` must return a `Permutation` representing the composition of `self` with `other`.  It is important to get this the correct way round: if `self` repsents $\sigma$ and `other` represents $\tau$ then the `Permutation` returned by `__mul__(self, other)` should represent $\sigma \circ \tau$, that is, `other` first, then `self`.
4. `inverse(self)` must return a new `Permutation` representing the inverse of the permutation represented by `self`.
5. `length(self)` must return the length of the permutation represeted by `self`. This is defined in [Algebra 1 coursework 2](https://moodle.ucl.ac.uk/mod/resource/view.php?id=5496462) question 13, or equivalently as the number of [inversions](https://en.wikipedia.org/wiki/Inversion_(discrete_mathematics)) of the permutation.
6. `disjoint_cycles(self, include_1_cycles=True)` must return a disjoint cycle decomposition of the permutation represented by `self`.  The method must return a list of lists in the form `[[a_1, ..., a_{n_1}], [b_1, ..., b_{n_2}], [c_1, ..., c_{n_3}], ...]` where $(a_1, \ldots, a_{n_1})(b_1, \ldots, b_{n_2})(c_1, \ldots, c_{n_3})\cdots$ is a disjoint cycle decomposition of the permutation represented by `self`.  The cycles can be ordered in any way, and you can start each cycle at any element of the cycle.  For example, if `s = Permutation([5, 8, 3, 1, 4, 7, 6, 2])` then `s` represents $(1, 5, 4)(2, 8)(3)(6, 7)$, so any of the following lists of lists would be correct outputs for `s.disjoint_cycles()`:

```python
[[1, 5, 4], [2, 8], [3], [6, 7]]
[[3], [5, 4, 1], [2, 8], [7, 6]]
[[6, 7], [8, 2], [4, 1, 5], [3]]
...
```

but `[[1, 4, 5], [2, 8], [3], [6, 7]]` would not be correct because $(1, 4, 5) \neq (1, 5, 4)$.
If the optional argument `include_1_cycles` is `False`, you must omit any 1-cycles from the final output, so in our example `[[1, 5, 4], [2, 8], [6, 7]]` would be one possible correct output.  You can compute disjoint cycle decompositions using the exact same method used in Algebra 1, but you will have to think carefully about how to implement it in Python.

Do not use the Permutation class from `sympy` or any other Python module.  You must write all of the methods yourselves.



````python
class Permutation:
    def __init__(self, bottom_row):
        """Create a Permutation. The argument bottom_row is the bottom row of the two row notation
        for the permutation being created"""
        self.bottom_row = bottom_row
    
    def __str__(self):
        bottom_row = '"'# creat an object to store the each element in the list in form of str 
        for i in self.bottom_row: #using for loop to turn every element into str
            bottom_row = bottom_row + (str(i)) +' ' #convert and add each element into (bottom_row) in                                                      form of str
        return bottom_row + '"' # return the permutation bottom row in str


    def __eq__(self, other):
        return self.bottom_row == other.bottom_row
        # it compares the bottom_row of self with the bottom_row of other and does Boolean           operation to give true false
    
    def __mul__(self, other):
        composition_bottom_row = [self.bottom_row[other.bottom_row[i] - 1] for i in range(len(self.bottom_row))]
        # new list called composition_bottom_row is created, it represents the bottom row         of the resulting permutation when self is composed with other. In the for loop,           for each index i in the bottom_row of self, it accesses an element in                     other.bottom_row[i], subtracts 1 from it to adjust for zero-based indexing, and           then uses this as an index to access an element in self.bottom_row.

        return Permutation(composition_bottom_row)
        
    
    def inverse(self):
        n = len(self.bottom_row)
        ip = [0] * n
        for i in range(1 , n+1):
            ip[i - 1] = self.bottom_row.index(i) +1
        return Permutation(ip)

    def length(self):
        #length of permutation = number of inversions
        total = 0 #reference point
        for i in range(len(self.bottom_row)): 
            for j in range(i+1,len(self.bottom_row)): #two for loops used to define two different                                                              numbers within the list
                if i<j and self.bottom_row[i] > self.bottom_row[j]: #to define what an inversion is
                    total+=1 #counter for each inversion within the permutation as the length is                                     essentially the number of inversions
        return total #put into same indent to obtain total number of inversions

    
    def disjoint_cycles(self, include_1_cycles=False):
        """Taking in a permutation as input, return a list containing lists representing disjoint                  cycles equivalent to the permutation"""
        l = len(self.bottom_row) #Length of the list representing the bottom row
        Top = [i for i in range(1, l + 1)] #Creating a list representing the top row
        d = [] #list containing the disjoint cycles
        while len(Top) != 0:
            a = Top[0] #takes the first element in the top row
            Top.remove(a) #remove the number from the list so we don't repeat a cycle
            n = self.bottom_row[a-1] #apply the permutation once
            c = [a] #List representing a single cycle
            while n != a:
                c.append(n)
                Top.remove(n)
                n = self.bottom_row[n-1]
                #apply the permutation until we get to the starting number, while appending the                         numbers to our list c and removing them from the top row list
            if include_1_cycles == False:
                if len(c) != 1:
                    d.append(c)
                    #if we choose to ommit all the 1-cycles, only add the cycle too the list of                             disjoint cycle if its length is longer than 1 
            else:
                 d.append(c)   
        
        return d
            
            
        
````

**Write a Python function `all_permutations(n)` which will return a list containing a `Permutation` object for every permutation in $S_n$.**  You can use the `permutations` function from [the `itertools` module](https://docs.python.org/3/library/itertools.html) to help - it will be much easier than generating the permutations yourself.

Don't call this for large values of `n` - it will take a long time and a lot of memory, because $|S_n| = n!$ grows very quickly.



````python
from itertools import permutations
def all_permutations(n):
    LOAP=[] #introduce a list to keep all permutation i generated. LOAP stand for List of all permutation 
    elements=[i for i in range (1,n+1)] # list of the element(1 to n), any permutation we are going to generate will have all this number once. 
    ps=permutations(elements, n) #ps is all permutation generated
    for p in ps: #we use for loop to converte each permutation into list (the input formate we need for our classes) and add each permutation into a list
        good_permutation=Permutation(list(p)) #list(p) convert p into a list formate, Permutation(list(p)), put list(p) into our classes to generate the permutaion in the formate we want.
        LOAP.append(good_permutation) #add those permutation we generate by our class into the final list
    return LOAP #return the final list
````

**Write a function `length_counts(n)` which returns a list whose `i`th entry is the number of permutations in $S_n$ with length $i$**  for each $0 \leqslant i \leqslant n(n-1)/2$.



````python
def length_counts(n):
    """Takes in an integer n as input, returns a list such that the i_th entry contains the number of          permutations of length i in Sn """
    S = all_permutations(n) #Create a list of all permutations in Sn
    L = int((n*(n - 1))/2) #Maximum length of the list to return
    C = [0] * (L+1) #Create a list of length L, initially with all 0s so that if permutations of length n                    do not exist, the n_th index returns 0
    for i in range(len(S)):
        n = S[i].length() #Takes the length of the permutation in Sn
        C[n] = C[n] + 1 #Add one to the n_th index everytime there is a permutation of length n
    
    return C
        
````

## Part 3: the $q$-factorial

The [`sympy` module](https://docs.sympy.org/latest/index.html) allows you to do symbolic calculations in Python.  For example, you can create a variable `q` using

````python
from sympy import *
q = Symbol('q')
````

and then create and manipulate algebraic *expressions* involving `q` by using ordinary Python arithmetic:



````python
f = 1+ 2*q + q**2
g = (1+q)**2
h = 1 + q + q + q**2
k = 1 + q**2 + 2*q
l = sum([q, q**2, 1, q])
````

These `sympy` expressions have a quite subtle notion of equality:

````python
print(f == g, f == h, f == k, f == l)
````

but it is easy to convert them to sympy `Poly` objects which makes them easier to compare for equality and easier to work with if we want to extract coefficients of powers of `q`, for example.

````python
fp = f.as_poly(q)
gp = g.as_poly(q)
hp = h.as_poly(q)
kp = k.as_poly(q)
lp = l.as_poly(q)
fp == gp == hp == kp == lp
````

For a positive integer $n$ we define $[n]_q = 1 + q + q^2 + \cdots + q^{n-1}$ and $[n]_q! = [1]_q[2]_q[3]_q \cdots [n]_q$, with the special cases $[0]_q! = 1$ and $[0]_q = 0$.  These are called $q$-analogues of the nonnegative integers and of the factorial function because if you set $q=1$ then $[n]_q$ becomes $1+1+\cdots +1 = n$ and $[n]_q^!$ becomes $1\cdot 2 \cdot 3 \cdots n = n!$.

**Write functions `q_integer(n)` and `q_factorial(n)` which return $[n]_q$ and $[n]_q!$** as expressions using the symbol `q` defined above.

You don't need to convert the output to a sympy `Poly` in the variable `q` yet.  Your function should work for all whole numbers $n \geqslant 0$.



````python
def q_integer(n):
    # we start with n=0, [0]q will be 0, this will be our initial strat value for sum in for loop 
    sum_q = 0
    # for loop to calculate the sum of the geometric series where q is the ratio
    for i in range(n):
        sum_q = sum_q + q**i
        # here what we are doing is we are updating the value of sum_q for everyiteration. sum_q will             be updated to its current value + new value (q**i) which will result in the new sum
    
    return sum_q # function returns the polynomial we want
````

````python
def q_factorial(n):
    # special case when n = 0 and 0[q]!=1, this will be our starting value for product in for loop
    product_n_q = 1
    # for loop to calculate the product of q_integers
    for i in range(1, n + 1):
        product_n_q = product_n_q * q_integer(i)
        # product_n_q 's value is updated in each iteration by being multiplied by the next q-integer           in the sequence.
    
    return product_n_q # function returns product of polynimials
````

**Write a function `q_factorial_coefficients(n)` which returns a list of length $1+ n(n-1)/2$ whose $i$th element is the coefficient of $q^i$ in $[n]_q!$.**  You can do this by converting `q_factorial(n)` into a sympy `Poly` in $q$ using `as_poly(q)`, but you'll need to **read the sympy documentation** which I linked to above to learn how to get the coefficients of a polynomial.

````python
def q_factorial_coefficients(n):
    q_fact_poly = poly(q_factorial(n),q) #converts the q-factorial into a single polynomial in terms of q
    coeff = q_fact_poly.all_coeffs() #extracts all of the coefficients from the polynomial
    
    return coeff[:] #prints out the coefficients into a list
````
